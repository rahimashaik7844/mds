```python
import pandas as pd

df = pd.read_csv('housep.csv')
print(df.head)
print(df.info())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
price_trend = df.groupby('YearBuilt')['Price'].mean().reset_index()

plt.figure(figsize=(10, 5))
plt.plot(price_trend['YearBuilt'], price_trend['Price'], color='blue', linewidth=2)

plt.title('Average House Price by Year Built')
plt.xlabel('Year Built')
plt.ylabel('Average Price ($)')
plt.grid(True)
plt.show()
```

```python
# barplot
plt.figure(figsize=(10, 5))
avg_price_location = df.groupby('Location')['Price'].mean().reset_index()

sns.barplot(data=avg_price_location, x='Location', y='Price', palette='Blues_d')

plt.title('Average House Price by Location')
plt.xlabel('Location')
plt.ylabel('Average Price ($)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
#density plot


# Density distribution of Price grouped by Condition
plt.figure(figsize=(10, 5))
sns.kdeplot(data=df, x='Price', hue='Condition', fill=True, common_norm=False, alpha=0.4)

plt.title('Price Density Distribution by Property Condition')
plt.xlabel('Price ($)')
plt.ylabel('Density')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
#histogram

# Distribution of House Area (Square Feet)
plt.figure(figsize=(10, 5))
plt.hist(df['Area'], bins=30, color='skyblue', edgecolor='black', alpha=0.8)

plt.title('Distribution of House Area (Square Feet)')
plt.xlabel('Area (Sq Ft)')
plt.ylabel('Frequency (Count)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
#scatter plot

# Area vs. Price colored by Garage availability
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Area', y='Price', hue='Garage', alpha=0.7, s=50)

plt.title('House Price vs. Area (Square Feet)')
plt.xlabel('Area (Sq Ft)')
plt.ylabel('Price ($)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
import matplotlib.cm as cm
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Load your dataset
df = pd.read_csv("housep.csv")

# Create figure and axis
fig, ax = plt.subplots(figsize=(14, 9), facecolor="#f8f9fa")  # Aesthetic 18

# Sample subset for clean marker visual separation
df_sample = df.sample(n=100, random_state=42).copy()

# -------------------------------------------------------------
# 1. Base Scatter Plot mapping 8 core aesthetics:
# -------------------------------------------------------------
scatter = sns.scatterplot(
    data=df_sample,
    x="Area",  # 1. X-position (Continuous)
    y="Price",  # 2. Y-position (Continuous)
    hue="Bedrooms",  # 3. Color Hue (Discrete/Numeric)
    size="Bathrooms",  # 4. Marker Size (Continuous)
    style="Location",  # 5. Marker Shape/Style (Categorical)
    alpha=0.8,  # 6. Transparency / Alpha
    sizes=(40, 240),
    palette="viridis",  # 7. Color Map / Palette
    ax=ax,
)

# -------------------------------------------------------------
# 2. Line Aesthetics & Error Bands:
# -------------------------------------------------------------
# Sort by Area for sequential line display
df_sorted = df_sample.sort_values("Area")

# 8. Line X, 9. Line Y, 10. Line Color, 11. Line Style, 12. Line Width
ax.plot(
    df_sorted["Area"],
    df_sorted["Price"].rolling(5).mean(),
    color="crimson",
    linestyle="--",
    linewidth=2.5,
    label="Price Trend (MA)",
)

# 13. Fill / Shading (Confidence Band)
ax.fill_between(
    df_sorted["Area"],
    df_sorted["Price"].rolling(5).mean() * 0.9,
    df_sorted["Price"].rolling(5).mean() * 1.1,
    color="red",
    alpha=0.15,
)

# -------------------------------------------------------------
# 3. Annotations, Markers, & Text Aesthetics:
# -------------------------------------------------------------
max_price_row = df_sample.loc[df_sample["Price"].idxmax()]

# 14. Annotation Arrow & Position
ax.annotate(
    f'Max Price: ${max_price_row["Price"]:,}',
    xy=(max_price_row["Area"], max_price_row["Price"]),
    xytext=(max_price_row["Area"] - 800, max_price_row["Price"] + 50000),
    arrowprops=dict(facecolor="black", shrink=0.05, width=1.5, headwidth=8),
    fontsize=10,  # 15. Text Font Size
    fontweight="bold",  # 16. Text Font Weight
    color="darkred",  # 17. Text Color
)

# -------------------------------------------------------------
# 4. Plot & Canvas Styling Aesthetics:
# -------------------------------------------------------------
ax.set_facecolor("#ffffff")  # 19. Axes/Panel Background Color

# 20. Gridline Style & 21. Gridline Color
ax.grid(True, which="major", linestyle=":", linewidth=0.8, color="gray", alpha=0.6)

# 22. Axis Spine Color / Border Visibility
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_color("black")
ax.spines["bottom"].set_color("black")

# 23. Title Formatting & Alignment
plt.title(
    "House Pricing: Demonstration of 23 Visual Aesthetics",
    fontsize=16,
    fontweight="bold",
    pad=20,
    loc="left",
)

plt.xlabel("Area (Sq Ft)", fontsize=11, fontweight="bold")
plt.ylabel("Price ($)", fontsize=11, fontweight="bold")
plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```
