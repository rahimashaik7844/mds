```python
# Option B: Fill missing values in specific numerical columns with Mean or Median
df_filled = df.copy()
df_filled['Study_Hours'] = df_filled['Study_Hours'].fillna(df_filled['Study_Hours'].median())
df_filled['Attendance(%)'] = df_filled['Attendance(%)'].fillna(df_filled['Attendance(%)'].mean())
df_filled['Grade'] = df_filled['Grade'].fillna(df_filled['Grade'].mode()[0])

df_filled.isnull().sum().sum()
```

```python
import pandas as pd
import numpy as np

# 1. Select the target column
col = 'Study_Hours'

# 2. Calculate Q1, Q3, and IQR
Q1 = df[col].quantile(0.25)
Q3 = df[col].quantile(0.75)
IQR = Q3 - Q1

# 3. Define lower and upper bounds
lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

# 4. Detect outliers (rows where value is outside bounds)
outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
print(f"Number of outliers in {col}:", len(outliers))
print(outliers)

# 5. Filter OUT outliers (keep only values within bounds)
df_clean = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]
```

```python
from scipy import stats

# 1. Calculate Z-scores for a column
z_scores = stats.zscore(df['Test_Score'])

# 2. Detect outliers (|Z-score| > 3)
outliers = df[abs(z_scores) > 3]
print("Outliers count:", len(outliers))

# 3. Filter OUT outliers
df_clean = df[abs(z_scores) <= 3]
```
