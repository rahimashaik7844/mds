```python
data={
    'Name':['Rahima','Sara','Ayesha'],
    'Age':[20,23,19],
    'Marks':[90,85,80]
}
df=pd.DataFrame(data)
print(df)
```

```python
data=[['Rahima',20,90],['Sara',23,85],['Ayesha',19,80]]
df=pd.DataFrame(data,columns=['Name','Age','Marks'])
print(df)
```

```python
name=pd.Series(['Rahima','Sara','Ayesha'])
age=pd.Series([20,23,19])
marks=pd.Series([90,85,80])
df=pd.DataFrame({'Name':name,'Age':age,'Marks':marks})
print(df)
```
